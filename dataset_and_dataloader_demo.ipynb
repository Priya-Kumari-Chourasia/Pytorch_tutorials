```python
from sklearn.datasets import make_classification
import torch
```

```python
X,y = make_classification(
    n_samples=10,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_classes=2,
    random_state=42
)
```

```python
X
```

```python
X=torch.tensor(X,dtype=torch.float32)
y=torch.tensor(y,dtype=torch.long)
```

```python
y
```

```python
from torch.utils.data import Dataset,DataLoader
```

```python
class cust(Dataset):
  def __init__(self,features,labels):
    self.features= features
    self.labels=labels

  def __len__(self):
    return self.features.shape[0]

  def __getitem__(self, index):
    return self.features[index],self.labels[index]
```

```python
dataset = cust(X,y)
```

```python
len(dataset)
```

```python
dataset[2]
```

```python
dataloader = DataLoader(dataset,batch_size=2,shuffle=True)
```

```python
for batch_features,batch_labels in dataloader:
  print(batch_features)
  print(batch_labels)
  print("-"*50)
```

collate_fn specifies how to combine a list of samples from a dataset into a single batch.by default the DataLoader uses a simple batch collation mechanism but collate_fn allows you to customize how the data should be processed  and  batched.
